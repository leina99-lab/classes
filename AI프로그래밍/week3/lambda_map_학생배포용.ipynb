# 1부: 익명 함수(lambda) — 이름 없이 즉석에서 만드는 함수

---

## lambda란 무엇인가?

**람다**(lambda)란 `def`로 정식 이름을 붙이지 않고 **한 줄로 즉석에서 만드는 임시 함수**이다.

```
[일반 함수 def]                    [람다 함수 lambda]

def add_ten(x):           →        lambda x : x + 10
    return x + 10

함수 이름이 있음                    이름이 없음 (익명)
여러 줄 가능                        한 줄만 가능
재사용 목적                         일회성 목적
```

---

### lambda의 구조 분해

```
lambda    x    :    x + 10
  ↑       ↑   ↑       ↑
예약어  매개변수 구분  표현식(실행할 연산)
             자
```

→ **`return` 키워드가 없다** — 표현식의 결과가 자동으로 반환된다.  
→ **조건문(`if/else`)도 한 줄로** 쓸 수 있다: `lambda x: '양수' if x > 0 else '음수'`

---

### AI 연결

```
[이미지 전처리]
normalize = lambda x: x / 255.0      # 픽셀값 0~1 정규화

[텍스트 전처리]
clean = lambda s: s.lower().strip()  # 소문자 + 공백 제거

[특성 변환]
log_scale = lambda x: np.log(x + 1) # 로그 스케일 변환

→ 이 세 줄이 실제 AI 코드에 그대로 들어간다.
```


### Ping 1 — `def` 함수와 `lambda` 함수 비교 (에어비앤비 가격 변환)

````python
# ─── Ping 1: def vs lambda ────────────────────────────────────
print("=" * 52)
print("  [방법 1] def 함수로 달러 → 원화 변환")
print("=" * 52)

def dollar_to_krw(dollar):
    return dollar * 1330

prices_usd = [80, 150, 220, 350, 420]
prices_krw_def = [dollar_to_krw(p) for p in prices_usd]

print(f"입력 (달러): {prices_usd}")
print(f"출력 (원화): {prices_krw_def}")
print()

print("─" * 52)
print("  [방법 2] lambda 함수로 동일한 변환")
print("─" * 52)

dollar_to_krw_lambda = lambda x: x * 1330

prices_krw_lambda = [dollar_to_krw_lambda(p) for p in prices_usd]

print(f"입력 (달러): {prices_usd}")
print(f"출력 (원화): {prices_krw_lambda}")
print()

print("─" * 52)
print("  [비교 정리]")
print("─" * 52)
print("  def   : 이름 있음 / 여러 줄 가능 / 재사용 목적")
print("  lambda: 이름 없음 / 한 줄만 가능 / 일회성 목적")
print()

print("─" * 52)
print("  [lambda 응용: 매개변수 여러 개]")
print("─" * 52)
add = lambda x, y: x + y
multiply = lambda x, y, z: x * y * z

print(f"add(3, 5)          = {add(3, 5)}")
print(f"multiply(2, 3, 4)  = {multiply(2, 3, 4)}")
print()

print("─" * 52)
print("  [lambda 응용: 조건식 if/else]")
print("─" * 52)
grade = lambda score: 'VIP' if score >= 100 else ('고급' if score >= 50 else '기본')

for r in [120, 80, 30]:
    print(f"리뷰 {r:3d}개 → 등급: {grade(r)}")

print()
print("[AI 연결]")
print("→ 위 grade 람다가 AI에서는 '활성화 함수'와 같은 역할을 한다.")
print("  ReLU = lambda x: max(0, x)  — 가장 유명한 활성화 함수")
relu = lambda x: max(0, x)
print(f"→ relu(-3) = {relu(-3)},  relu(5) = {relu(5)}")
````

### ✎ 개념 확인 1

**Q1.** `lambda x: x * 2` 에서 `return` 키워드가 없어도 값이 반환되는 이유는 무엇인가?

**Q2.** `def`와 `lambda` 중 재사용 가능한 복잡한 함수에는 어떤 것을 써야 하는가?

**Q3.** `relu = lambda x: max(0, x)` 에서 `relu(-5)` 와 `relu(3)` 의 결과는?

<details>
<summary>▶ 정답 보기</summary>

**A1.** `lambda`는 **표현식**(expression)의 계산 결과를 자동으로 반환하도록 설계되어 있다. `def`에서는 명시적으로 `return`을 써야 하지만, `lambda`에서는 표현식 자체가 반환값이다.

**A2.** `def`를 사용해야 한다. `lambda`는 한 줄 표현식만 가능하고, 코드 가독성을 위해 복잡한 로직에는 적합하지 않다. PEP 8 스타일 가이드에서도 람다를 변수에 대입하는 방식은 지양하도록 권고한다.

**A3.** `relu(-5) = 0`,  `relu(3) = 3`. 음수는 0으로, 양수는 그대로 통과시킨다. 이것이 딥러닝에서 가장 많이 쓰이는 활성화 함수 ReLU(Rectified Linear Unit)이다.

</details>

# 2부: 일괄 처리 `map()` — 전체에 한 번에 적용하는 공장 컨베이어 벨트

---

## `map()`이란 무엇인가?

**`map()`** 은 리스트(또는 반복 가능한 객체)의 **모든 요소에 함수를 일괄 적용**하는 고차 함수이다.

```
[공장 컨베이어 벨트 비유]

원재료 창고       컨베이어 벨트        완성품 창고
[80, 150, 220]  →  ×1330 기계  →  [106400, 199500, 292600]
   (달러)          (함수)               (원화)
     ↑                ↑                    ↑
   리스트           함수             map 객체 → list()
```

---

### `map()` 작동 원리 (3단계)

```
Step 1: 리스트에서 요소를 하나씩 꺼낸다
        [80,  150,  220]
          ↓    ↓    ↓

Step 2: 꺼낸 요소를 함수에 대입한다
        f(80)  f(150)  f(220)
          ↓      ↓      ↓

Step 3: 결과를 모아 map 객체로 반환
        map object → list([106400, 199500, 292600])
```

---

### `map()` 기본 문법

```python
map(적용할_함수,  반복_가능한_객체)
      ↑                ↑
   lambda 또는      리스트, 튜플,
   def 함수         Series 등
```

→ 반환값은 `map 객체`이므로 **`list()`로 감싸서** 결과를 확인한다.  
→ 판다스에서는 `Series.map()` 또는 `df['열'].apply()`로 동일하게 작동한다.


### Ping 2 — `map()` 기본 사용법 (에어비앤비 가격 변환)

````python
# ─── Ping 2: map() 기본 사용법 ────────────────────────────────
prices_usd = [80, 150, 220, 350, 420, 95, 65, 290, 130, 180]
숙소명 = ['A','B','C','D','E','F','G','H','I','J']

print("=" * 52)
print("  [Step 1] map 객체의 정체 확인")
print("=" * 52)
raw_map = map(lambda x: x * 1330, prices_usd)
print(f"map() 반환값 타입: {type(raw_map)}")
print(f"list()로 변환   : {list(raw_map)}")
print()

print("─" * 52)
print("  [Step 2] 다양한 변환 일괄 적용")
print("─" * 52)

# 달러 → 원화
prices_krw = list(map(lambda x: x * 1330, prices_usd))

# 달러 → 유로 (1달러 ≈ 0.92유로)
prices_eur = list(map(lambda x: round(x * 0.92, 1), prices_usd))

# 가격 등급 분류
price_grade = list(map(lambda x: '저가' if x < 100 else ('중가' if x < 250 else '고가'),
                        prices_usd))

print(f"{'숙소':>4} | {'달러($)':>7} | {'원화(₩)':>9} | {'유로(€)':>7} | 등급")
print("─" * 48)
for name, usd, krw, eur, g in zip(숙소명, prices_usd, prices_krw, prices_eur, price_grade):
    print(f" 숙소{name}  | {usd:>7,} | {krw:>9,} | {eur:>7,.1f} | {g}")
print()

print("─" * 52)
print("  [Step 3] def 함수와 함께 사용")
print("─" * 52)

def categorize_price(p):
    if p < 100:
        return f"${p:3d} → 저가 (예산 여행자 추천)"
    elif p < 250:
        return f"${p:3d} → 중가 (일반 여행자 추천)"
    else:
        return f"${p:3d} → 고가 (럭셔리 여행자 추천)"

descriptions = list(map(categorize_price, prices_usd))
for d in descriptions:
    print(f"  {d}")
print()
print("[AI 연결]")
print("→ map()은 내부적으로 C 언어로 최적화되어 있어")
print("  for 반복문보다 대용량 데이터에서 더 빠르게 작동한다.")
print("→ 딥러닝에서 배치(batch) 처리와 동일한 개념이다.")
````

### ✎ 개념 확인 2

**Q1.** `map()` 의 반환값이 리스트가 아닌 `map 객체`인 이유는 무엇인가?

**Q2.** `map(함수, 리스트)` 에서 리스트 대신 다른 어떤 자료형을 쓸 수 있는가?

**Q3.** 딥러닝의 '배치(batch) 처리'와 `map()`의 공통점은 무엇인가?

<details>
<summary>▶ 정답 보기</summary>

**A1.** `map()`은 **지연 평가**(Lazy Evaluation)를 사용한다. 즉, `list()`로 변환하거나 값이 실제로 필요할 때까지 계산을 미룬다. 이는 수억 개의 요소가 있을 때 메모리를 절약하기 위한 설계이다.

**A2.** 리스트뿐 아니라 **튜플, 문자열, 집합, range, 판다스 Series** 등 반복 가능한(iterable) 모든 객체를 사용할 수 있다.

**A3.** 둘 다 **여러 데이터에 동일한 연산을 병렬적으로 일괄 적용**한다는 원리가 같다. 딥러닝의 배치 처리는 32개, 64개, 128개 샘플을 동시에 연산하는데, 이는 `map()`이 리스트의 모든 요소에 함수를 동시에 적용하는 것과 같은 개념이다.

</details>


### Pong 2 — `map()` 일괄 처리 (COVID-19 데이터)

**미션:** `map()`을 사용하여 COVID-19 확진자 데이터를 일괄 변환하라.

| 변환 | 규칙 |
|:---|:---|
| 만 명 단위 변환 | 확진자 수 ÷ 10,000 (소수 1자리) |
| 위험도 분류 | 1,000만 미만 → '관리', 1,000만~5,000만 → '경계', 5,000만 이상 → '심각' |
| 로그 변환 | `np.log10(확진자)` 로 변환 (소수 2자리) |


````python
# ─── Pong 2: 여기에 코드를 작성하시오 ─────────────────────────
import numpy as np

covid_cases = [4_500_000, 95_000_000, 44_000_000, 35_000_000, 23_000_000]
countries   = ['대한민국', '미국', '인도', '브라질', '영국']

# 1. 만 명 단위 변환
to_man = list(map(___, covid_cases))

# 2. 위험도 분류 (1000만 미만: '관리', ~5000만: '경계', 이상: '심각')
danger = list(map(___, covid_cases))

# 3. 로그 변환
log_vals = list(map(___, covid_cases))

# 결과 출력
print("=" * 60)
print(f"{'나라':>6} | {'확진(만명)':>10} | {'위험도':>6} | {'log10':>6}")
print("─" * 60)
for c, m, d, l in zip(countries, to_man, danger, log_vals):
    print(f"{c:>6} | {m:>10.1f} | {d:>6} | {l:>6.2f}")
````

# 3부: 종합 응용 — `map` + `lambda` vs `for`문

---

## 핵심 비교: 왜 `map + lambda`가 더 좋은가?

```
[for 반복문 방식]                [map + lambda 방식]

result = []                    result = list(
for c in covid_cases:    →         map(lambda x: x/10000,
    result.append(              covid_cases))
        c / 10000)

4줄, 명령형 코드               1줄, 선언형 코드
"어떻게 해라"                  "무엇을 해라"
```

---

### 세 가지 방식 속도 비교 (100만 개 데이터)

```
for 반복문:          ████████████████████  0.15초
list comprehension:  ██████████            0.08초
map + lambda:        ████████              0.06초  ← 가장 빠름
```

→ 데이터가 클수록 `map()`의 속도 우위가 더 커진다.  
→ AI 전처리에서 수백만~수천만 개의 데이터를 다룰 때 이 차이가 결정적이다.


### Ping 3 — 세 가지 방식 완전 비교 (에어비앤비 데이터)

````python
# ─── Ping 3: for / list comprehension / map+lambda 비교 ───────
import time

prices_usd = [80, 150, 220, 350, 420, 95, 65, 290, 130, 180,
              200, 310, 75, 160, 240, 120, 185, 275, 90, 330]

print("=" * 52)
print("  [방법 1] for 반복문 (전통 방식)")
print("=" * 52)
t0 = time.perf_counter()

result_for = []
for p in prices_usd:
    result_for.append(p * 1330)

t1 = time.perf_counter()
print(f"코드 줄 수: 3줄")
print(f"결과: {result_for[:5]} ...")
print(f"소요 시간: {(t1-t0)*1e6:.2f} 마이크로초")
print()

print("─" * 52)
print("  [방법 2] 리스트 컴프리헨션")
print("─" * 52)
t0 = time.perf_counter()

result_comp = [p * 1330 for p in prices_usd]

t1 = time.perf_counter()
print(f"코드 줄 수: 1줄")
print(f"결과: {result_comp[:5]} ...")
print(f"소요 시간: {(t1-t0)*1e6:.2f} 마이크로초")
print()

print("─" * 52)
print("  [방법 3] map + lambda (추천)")
print("─" * 52)
t0 = time.perf_counter()

result_map = list(map(lambda p: p * 1330, prices_usd))

t1 = time.perf_counter()
print(f"코드 줄 수: 1줄")
print(f"결과: {result_map[:5]} ...")
print(f"소요 시간: {(t1-t0)*1e6:.2f} 마이크로초")
print()

print("─" * 52)
print("  [고급] 두 리스트를 동시에 처리 — map + 두 인자")
print("─" * 52)
prices_list     = [80, 150, 220, 350, 420]
min_nights_list = [2,  3,   1,   5,   2  ]

# 최소 숙박 총비용 계산
total_costs = list(map(lambda p, n: p * n, prices_list, min_nights_list))
print("최소 숙박 총비용 (가격 × 최소숙박일):")
for p, n, t in zip(prices_list, min_nights_list, total_costs):
    print(f"  ${p:3d} × {n}박 = ${t:4d}")
print()

print("[AI 연결]")
print("→ 두 리스트를 동시에 처리하는 map(f, list1, list2)은")
print("  딥러닝에서 예측값과 실제값을 동시에 받아 손실을 계산하는")
print("  loss = criterion(pred, target) 과 같은 구조이다.")
````

## 판다스에서 lambda를 쓰는 세 가지 방법

```
[방법 1] Series.map()        → 열의 각 값에 함수 적용
[방법 2] Series.apply()      → 열의 각 값에 함수 적용 (더 강력)
[방법 3] DataFrame.apply()   → 행 또는 열 전체에 함수 적용

→ 셋 다 내부적으로 map()과 동일한 원리로 작동한다.
```

---

### AI 파이프라인에서의 위치

```
판다스 DataFrame
    ↓  df['가격'].apply(lambda x: np.log(x+1))  ← 오늘 배우는 것
로그 변환된 특성
    ↓  df['동네'].map({'Manhattan':0, 'Brooklyn':1})
레이블 인코딩
    ↓  (X - X.mean()) / X.std()
표준화된 특성 행렬 X
    ↓  torch.tensor(X.values)
AI 모델 입력
```


### Ping 4 — 에어비앤비 데이터에 `apply()` 적용하기

````python
# ─── Ping 4: 판다스 apply() ───────────────────────────────────
AIRBNB_URL = (
    "https://raw.githubusercontent.com/leina99-lab/classes/main/"
    "AI%ED%94%84%EB%A1%9C%EA%B7%B8%EB%9E%98%EB%B0%8D/data/AB_NYC_2019.csv"
)

try:
    df = pd.read_csv(AIRBNB_URL)
    df = df[df['price'] > 0].dropna(subset=['price', 'neighbourhood_group'])
    print(f"→ 에어비앤비 데이터 로딩 성공! 형태: {df.shape}")
except Exception:
    import random
    random.seed(42)
    n = 300
    groups = ['Manhattan','Brooklyn','Queens','Bronx','Staten Island']
    room_types = ['Entire home/apt','Private room','Shared room']
    df = pd.DataFrame({
        'id'                  : [10000 + i*11 for i in range(n)],
        'neighbourhood_group' : [groups[i % 5] for i in range(n)],
        'room_type'           : [room_types[i % 3] for i in range(n)],
        'price'               : [30 + (i*53) % 470 for i in range(n)],
        'number_of_reviews'   : [(i*17) % 200 for i in range(n)],
        'minimum_nights'      : [1 + i % 15 for i in range(n)],
    })
    print(f"→ [대체 데이터] 합성 에어비앤비 데이터 생성. 형태: {df.shape}")

# 분석용 열만 선택
cols = ['neighbourhood_group', 'room_type', 'price', 'number_of_reviews', 'minimum_nights']
cols = [c for c in cols if c in df.columns]
sample = df[cols].head(8).copy()

print()
print("─" * 55)
print("  원본 데이터 (상위 8행)")
print("─" * 55)
print(sample.to_string(index=True))
print()

print("=" * 55)
print("  [apply 1] 가격 등급 열 추가")
print("=" * 55)
sample['가격_등급'] = sample['price'].apply(
    lambda x: '저가' if x < 100 else ('중가' if x < 250 else '고가')
)
print(sample[['price', '가격_등급']].to_string())
print()

print("─" * 55)
print("  [apply 2] 원화 환산 열 추가")
print("─" * 55)
sample['가격_원화'] = sample['price'].apply(lambda x: f"₩{x * 1330:,}")
print(sample[['price', '가격_원화']].to_string())
print()

print("─" * 55)
print("  [map 3] 동네 → 숫자 레이블 변환 (Label Encoding)")
print("─" * 55)
label_map = {
    'Manhattan': 0, 'Brooklyn': 1, 'Queens': 2,
    'Bronx': 3, 'Staten Island': 4
}
if 'neighbourhood_group' in sample.columns:
    sample['동네_코드'] = sample['neighbourhood_group'].map(label_map)
    print(sample[['neighbourhood_group', '동네_코드']].to_string())
    print()
    print("[AI 연결]")
    print("→ 이것이 바로 '레이블 인코딩(Label Encoding)'이다.")
    print("→ AI 모델은 문자열을 처리할 수 없어서 숫자로 변환해야 한다.")
    print("→ sklearn의 LabelEncoder가 내부적으로 이와 동일한 작업을 수행한다.")
````

**Q3.** 레이블 인코딩(Label Encoding)이 AI 모델에 반드시 필요한 이유는 무엇인가?

<details>
<summary>▶ 정답 보기</summary>

**A1.** `map()`은 주로 **딕셔너리나 함수**로 값을 교체/변환하는 데 사용하며, `apply()`는 더 복잡한 **임의의 함수(lambda 포함)**를 적용할 수 있다. 단순 변환은 `map()`, 복잡한 연산은 `apply()`를 권장한다.

**A2.** 딕셔너리에 없는 키에 해당하는 값은 `NaN`(결측값)으로 처리된다. 이를 방지하려면 `.fillna()`로 후처리하거나, `lambda`와 `.get()` 메서드를 결합하면 된다: `df['동네'].map(lambda x: label_map.get(x, -1))`.

**A3.** 모든 AI 모델(신경망, 랜덤포레스트, SVM 등)은 **수치형 데이터만 입력**으로 받는다. 문자열('Manhattan')을 직접 입력하면 오류가 발생하므로, 반드시 숫자로 변환(인코딩)해야 한다. 레이블 인코딩은 가장 단순한 인코딩 방법이다.

## 오늘의 핵심 공식

| 구문 | 역할 | AI 사용 예 |
|:---|:---|:---|
| `lambda x: x * 1330` | 익명 단일 변환 | 픽셀 정규화 `lambda x: x/255` |
| `lambda x, y: x + y` | 다중 인자 처리 | 손실 계산 `lambda pred, y: (pred-y)**2` |
| `lambda x: A if 조건 else B` | 조건 분기 | ReLU `lambda x: max(0, x)` |
| `list(map(f, lst))` | 리스트 일괄 변환 | 전처리 파이프라인 |
| `list(map(f, l1, l2))` | 두 리스트 동시 처리 | 예측값·정답 동시 처리 |
| `df['열'].apply(lambda x: ...)` | 판다스 열 변환 | 특성 엔지니어링 |
| `df['열'].map(딕셔너리)` | 레이블 인코딩 | 범주형 → 수치형 |
| `df.apply(f, axis=1)` | 행 단위 함수 적용 | 복합 특성 생성 |

---

## 전체 AI 파이프라인에서의 위치

```
원본 CSV 데이터
        ↓  pd.read_csv()
판다스 DataFrame
        ↓  df['열'].apply(lambda x: ...)    ← 오늘 배운 것
특성 변환 & 레이블 인코딩
        ↓  df['열'].map({'A':0, 'B':1})      ← 오늘 배운 것
전처리 완료 DataFrame
        ↓  df.values
넘파이 배열
        ↓  torch.tensor()
파이토치 텐서 → AI 모델 훈련
```
