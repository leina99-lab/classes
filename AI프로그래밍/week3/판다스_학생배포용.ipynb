## 시리즈(Series)란 무엇인가?

**시리즈**(Series)는 **인덱스(이름표)가 붙은 1차원 배열**이다.

```
파이썬 리스트:             판다스 시리즈:
[150, 220, 80, 350]   →   숙소A    150
                          숙소B    220
                          숙소C     80
                          숙소D    350
                          Name: 1박_가격_달러, dtype: int64
```

→ **리스트와의 차이**: 각 값에 **인덱스**(이름표)가 붙어 있어 이름으로 접근할 수 있다.  
→ **AI 관점**: 신경망에 입력되는 하나의 **특성(feature) 벡터**가 바로 시리즈 형태이다.

---

### 시리즈의 구조 (시각화)

```
      Index(인덱스)    Value(값)
         숙소A    →    150      ←─ 1박 가격(달러)
         숙소B    →    220
         숙소C    →     80
         숙소D    →    350
            ↑              ↑
        이름표          실제 데이터
```


### Ping 1 — 시리즈 생성과 기본 속성 (에어비앤비 가격 데이터)

에어비앤비 NYC 숙소 10곳의 1박 가격으로 시리즈를 만들어본다.

````python
# ─── Ping 1: Series 생성 ───────────────────────────────────────
prices = pd.Series(
    [150, 220, 80, 350, 95, 180, 420, 65, 290, 130],
    index=['숙소A', '숙소B', '숙소C', '숙소D', '숙소E',
           '숙소F', '숙소G', '숙소H', '숙소I', '숙소J'],
    name='1박_가격_달러'
)

print("=" * 48)
print("  에어비앤비 NYC 숙소 가격 시리즈")
print("=" * 48)
print(prices)
print()
print(f"→ 타입     : {type(prices)}")
print(f"→ 크기     : {len(prices)} 개")
print(f"→ 데이터형  : {prices.dtype}")
print(f"→ 시리즈명  : {prices.name}")
print(f"→ 인덱스   : {prices.index.tolist()}")
print()
print("─" * 48)
print("  [기술통계량]")
print("─" * 48)
print(f"  최솟값  : $ {prices.min()}")
print(f"  최댓값  : $ {prices.max()}")
print(f"  평균    : $ {prices.mean():.1f}")
print(f"  중앙값  : $ {prices.median()}")
print(f"  합계    : $ {prices.sum()}")
print()
print("[AI 연결]")
print("→ 이 10개의 가격값이 신경망에 입력되는")
print("  하나의 특성(feature) 벡터이다.")
print("→ 신경망은 이런 벡터 수십만 개를 동시에 처리한다.")
````

### Ping 2 — 시리즈 인덱싱과 슬라이싱

시리즈에서 원하는 값을 꺼내는 5가지 방법을 배운다.

````python
# ─── Ping 2: Series 인덱싱 & 슬라이싱 ────────────────────────
prices = pd.Series(
    [150, 220, 80, 350, 95, 180, 420, 65, 290, 130],
    index=['숙소A','숙소B','숙소C','숙소D','숙소E',
           '숙소F','숙소G','숙소H','숙소I','숙소J'],
    name='1박_가격_달러'
)

print("=" * 48)
print("  [방법 1] 인덱스 이름으로 접근")
print("=" * 48)
print(f"숙소A 가격: $ {prices['숙소A']}")
print(f"숙소D 가격: $ {prices['숙소D']}")

print()
print("─" * 48)
print("  [방법 2] 위치 번호로 접근 (.iloc)")
print("─" * 48)
print(f"첫 번째 숙소 : $ {prices.iloc[0]}")
print(f"마지막 숙소  : $ {prices.iloc[-1]}")
print(f"세 번째 숙소 : $ {prices.iloc[2]}")

print()
print("─" * 48)
print("  [방법 3] 여러 개 동시 선택")
print("─" * 48)
print(prices[['숙소A', '숙소C', '숙소G']])

print()
print("─" * 48)
print("  [방법 4] 슬라이싱")
print("─" * 48)
print("처음 3개:")
print(prices[:3])
print()
print("뒤에서 3개:")
print(prices[-3:])

print()
print("─" * 48)
print("  [방법 5] 조건 필터링 (Boolean Indexing)")
print("─" * 48)
cheap = prices[prices < 100]
print(f"100달러 미만 숙소:\n{cheap}")
print()
expensive = prices[prices >= 300]
print(f"300달러 이상 숙소:\n{expensive}")

print()
print("[AI 연결]")
print("→ 조건 필터링 = AI 훈련 시 특정 샘플만 선택하는")
print("  '마스킹(Masking)' 기법의 기초이다.")
print("→ 트랜스포머(Transformer)의 어텐션 마스크도 같은 원리이다.")
````

### Ping 3 — 시리즈 연산과 주요 메서드

시리즈끼리 연산하고, 유용한 메서드를 익힌다.

````python
# ─── Ping 3: Series 연산 & 주요 메서드 ──────────────────────
prices = pd.Series(
    [150, 220, 80, 350, 95, 180, 420, 65, 290, 130],
    index=['숙소A','숙소B','숙소C','숙소D','숙소E',
           '숙소F','숙소G','숙소H','숙소I','숙소J'],
    name='1박_가격_달러'
)

print("=" * 48)
print("  [산술 연산] 벡터화(Vectorization)")
print("=" * 48)
prices_krw = prices * 1330   # 달러 → 원화 (1달러 ≈ 1330원)
print("원화 환산 가격 (1달러 = 1330원):")
print(prices_krw)
print()

print("─" * 48)
print("  [정렬] sort_values()")
print("─" * 48)
print("오름차순 (싼 순):")
print(prices.sort_values())
print()
print("내림차순 (비싼 순):")
print(prices.sort_values(ascending=False))
print()

print("─" * 48)
print("  [주요 메서드 한눈에]")
print("─" * 48)
print(f"  count()      : {prices.count()}")
print(f"  nunique()    : {prices.nunique()}")
print(f"  quantile(25%): $ {prices.quantile(0.25)}")
print(f"  quantile(50%): $ {prices.quantile(0.50)}")
print(f"  quantile(75%): $ {prices.quantile(0.75)}")
print()

print("─" * 48)
print("  [한번에 보기: describe()]")
print("─" * 48)
print(prices.describe())
print()

print("─" * 48)
print("  [AI 연결] 정규화(Normalization)")
print("─" * 48)
print("→ AI 모델에 넣기 전 0~1 사이로 변환:")
normalized = (prices - prices.min()) / (prices.max() - prices.min())
print(normalized.round(3))
print()
print("→ 정규화 공식: (값 - 최솟값) / (최댓값 - 최솟값)")
print("→ 이 공식이 딥러닝 전처리의 기본 중의 기본이다.")
````

## 데이터프레임(DataFrame)이란 무엇인가?

**데이터프레임**(DataFrame)은 **여러 개의 시리즈가 열(Column)로 결합된 2차원 표**이다.

```
         이름          동네    방_유형  1박_가격  리뷰_수
행(0)  Cozy Studio   Brooklyn   전체     150      45
행(1)  Modern Loft   Manhattan  전체     220      82
행(2)  Budget Room   Queens    개인실    80       12
행(3)  Luxury Suite  Manhattan  전체     350      156
   ↑         ↑           ↑         ↑        ↑        ↑
인덱스        열(Column) 각각이 하나의 Series
```

→ 엑셀의 **시트(Sheet)**와 구조가 완전히 동일하다.  
→ AI 관점: 딥러닝 훈련 데이터셋의 기본 단위가 데이터프레임 형태이다.


### Ping 4 — 데이터프레임 생성과 기본 정보 확인

````python
# ─── Ping 4: DataFrame 생성 ────────────────────────────────────
airbnb_sample = pd.DataFrame({
    '숙소명'     : ['Cozy Studio', 'Modern Loft', 'Budget Room',
                    'Luxury Suite', 'Tiny House'],
    '동네'       : ['Brooklyn', 'Manhattan', 'Queens',
                    'Manhattan', 'Brooklyn'],
    '방_유형'    : ['전체', '전체', '개인실', '전체', '전체'],
    '1박_가격'   : [150, 220, 80, 350, 95],
    '최소_숙박일' : [2, 3, 1, 5, 2],
    '리뷰_수'    : [45, 82, 12, 156, 28]
})

print("=" * 58)
print("  에어비앤비 NYC 샘플 데이터프레임")
print("=" * 58)
print(airbnb_sample)
print()

print("─" * 58)
print("  [기본 속성]")
print("─" * 58)
print(f"→ 형태 (행×열)   : {airbnb_sample.shape}")
print(f"→ 행 수          : {airbnb_sample.shape[0]} 개")
print(f"→ 열 수          : {airbnb_sample.shape[1]} 개")
print(f"→ 열 이름들      : {list(airbnb_sample.columns)}")
print(f"→ 인덱스         : {list(airbnb_sample.index)}")
print()

print("─" * 58)
print("  [데이터 타입: .dtypes]")
print("─" * 58)
print(airbnb_sample.dtypes)
print()

print("─" * 58)
print("  [전체 요약: .info()]")
print("─" * 58)
airbnb_sample.info()
print()
print("[AI 연결]")
print("→ object 타입 열 = 문자열 → AI 입력 전 숫자로 변환 필요")
print("  (One-Hot Encoding, Label Encoding 등)")
print("→ int64 타입 열  = 바로 AI 모델에 입력 가능")
````

### Ping 5 — 행/열 선택: loc 과 iloc

DataFrame에서 원하는 데이터를 정밀하게 추출하는 핵심 문법을 배운다.

````python
# ─── Ping 5: loc / iloc ────────────────────────────────────────
airbnb_sample = pd.DataFrame({
    '숙소명'     : ['Cozy Studio','Modern Loft','Budget Room',
                    'Luxury Suite','Tiny House'],
    '동네'       : ['Brooklyn','Manhattan','Queens','Manhattan','Brooklyn'],
    '방_유형'    : ['전체','전체','개인실','전체','전체'],
    '1박_가격'   : [150, 220, 80, 350, 95],
    '최소_숙박일' : [2, 3, 1, 5, 2],
    '리뷰_수'    : [45, 82, 12, 156, 28]
})

print("=" * 58)
print("  [열(Column) 선택]")
print("=" * 58)
print("▷ 단일 열 — df['열이름']:")
print(airbnb_sample['숙소명'])
print()
print("▷ 여러 열 — df[['열1','열2']]:")
print(airbnb_sample[['1박_가격', '리뷰_수']])
print()

print("─" * 58)
print("  [.loc — 이름으로 행/열 선택]")
print("─" * 58)
print("▷ 0번 행 전체:")
print(airbnb_sample.loc[0])
print()
print("▷ 0~2행, 숙소명과 가격만:")
print(airbnb_sample.loc[0:2, ['숙소명', '1박_가격']])
print()

print("─" * 58)
print("  [.iloc — 위치 번호로 행/열 선택]")
print("─" * 58)
print("▷ 첫 번째 행:")
print(airbnb_sample.iloc[0])
print()
print("▷ 처음 3행, 처음 3열:")
print(airbnb_sample.iloc[:3, :3])
print()

print("─" * 58)
print("  [AI 연결] X와 y 분리")
print("─" * 58)
print("→ AI 모델 학습 시 특성과 정답을 분리하는 코드:")
print("  X = df.iloc[:, :-1]  # 마지막 열 제외 = 특성(feature)")
print("  y = df.iloc[:, -1]   # 마지막 열 = 정답(label)")
print()
X = airbnb_sample.iloc[:, 3:]   # 수치 열만
y = airbnb_sample.iloc[:, 0]    # 숙소명
print(f"특성 X 형태: {X.shape}  →  정답 y 형태: {y.shape}")
````

### ✎ 개념 확인 5

**Q1.** `.loc` 과 `.iloc` 의 핵심 차이는 무엇인가?

**Q2.** `df.iloc[:, :-1]` 에서 `:-1` 은 무슨 뜻인가?

**Q3.** AI 모델 학습에서 X와 y를 반드시 분리해야 하는 이유는 무엇인가?

<details>
<summary>▶ 정답 보기</summary>

**A1.** `.loc`은 인덱스 **이름**으로 접근하고, `.iloc`은 **정수 위치**로 접근한다. 인덱스 이름이 숫자가 아닌 문자열일 때 특히 차이가 명확하다.

**A2.** 처음부터 마지막 열을 제외한 모든 열을 선택한다. 마지막 열이 보통 정답(label) 열이므로, 이 구문으로 특성(feature)만 골라낸다.

**A3.** X는 모델에 입력되는 특성(feature)이고, y는 모델이 맞춰야 할 정답(label)이다. 모델이 정답을 미리 보고 학습하면 진짜 예측 능력을 측정할 수 없다.

</details>

# 3부: 실제 데이터 불러오기 & 탐색적 데이터 분석(EDA)

## 탐색적 데이터 분석(EDA)이란?

**탐색적 데이터 분석**(Exploratory Data Analysis, EDA)이란 데이터를 다양한 각도로 살펴보며 **패턴, 이상치, 관계**를 발견하는 과정이다.

```
원본 데이터
    ↓  1단계: 크기·형태 확인    → .shape, .info()
    ↓  2단계: 기술통계 확인     → .describe()
    ↓  3단계: 결측값 확인       → .isnull().sum()
    ↓  4단계: 분포 확인         → .value_counts(), 시각화
    ↓  5단계: 상관관계 확인     → .corr()
AI 모델 학습 준비 완료
```

→ EDA는 AI 프로젝트 전체 시간의 **60~80%** 를 차지하는 가장 중요한 단계이다.  
→ 데이터를 모르면 모델도 모른다 — **"Garbage In, Garbage Out"**


### Ping 6 — 에어비앤비 데이터 불러오기 (48,895개 실제 데이터)

````python
# ─── Ping 6: CSV 불러오기 ──────────────────────────────────────
AIRBNB_URL = (
    "https://raw.githubusercontent.com/leina99-lab/classes/main/"
    "AI%ED%94%84%EB%A1%9C%EA%B7%B8%EB%9E%98%EB%B0%8D/data/AB_NYC_2019.csv"
)

try:
    airbnb = pd.read_csv(AIRBNB_URL)
    print("→ 에어비앤비 데이터 로딩 성공!")
    print(f"→ 데이터 형태: {airbnb.shape}")
except Exception:
    import random
    random.seed(42)
    n = 500
    neighbourhoods = ['Brooklyn', 'Manhattan', 'Queens', 'Bronx', 'Staten Island']
    room_types = ['Entire home/apt', 'Private room', 'Shared room']
    airbnb = pd.DataFrame({
        'id'                            : [10000 + i*7 for i in range(n)],
        'name'                          : [f'Unique_Place_{i*13}' for i in range(n)],
        'neighbourhood_group'           : [neighbourhoods[i % 5] for i in range(n)],
        'neighbourhood'                 : [f'Area_{(i*3) % 30 + 1}' for i in range(n)],
        'room_type'                     : [room_types[i % 3] for i in range(n)],
        'price'                         : [30 + (i*47) % 470 for i in range(n)],
        'minimum_nights'                : [1 + i % 10 for i in range(n)],
        'number_of_reviews'             : [(i*11) % 200 for i in range(n)],
        'availability_365'              : [(i*7) % 365 for i in range(n)],
        'calculated_host_listings_count': [1 + i % 20 for i in range(n)],
    })
    print("→ [대체 데이터] 합성 에어비앤비 데이터 생성 완료")
    print(f"→ 데이터 형태: {airbnb.shape}")

print()
print("─" * 55)
print("  처음 5행 미리보기 (.head())")
print("─" * 55)
print(airbnb.head())
````

### Ping 7 — EDA: 기본 탐색 함수 총정리

````python
# ─── Ping 7: EDA 기본 탐색 ────────────────────────────────────
print("=" * 55)
print("  [Step 1] 크기와 구조 파악")
print("=" * 55)
print(f"→ 행(숙소 수): {airbnb.shape[0]:,} 개")
print(f"→ 열(특성 수): {airbnb.shape[1]} 개")
print(f"→ 열 이름    : {list(airbnb.columns)}")
print()

print("─" * 55)
print("  [Step 2] 앞/뒤 확인")
print("─" * 55)
print("▷ head(3):")
print(airbnb.head(3))
print()
print("▷ tail(3):")
print(airbnb.tail(3))
print()

print("─" * 55)
print("  [Step 3] 타입 & 결측값: .info()")
print("─" * 55)
airbnb.info()
print()

print("─" * 55)
print("  [Step 4] 기술통계: .describe()")
print("─" * 55)
print(airbnb.describe().round(2))
print()

print("─" * 55)
print("  [Step 5] 결측값 집중 점검")
print("─" * 55)
missing = airbnb.isnull().sum()
missing_pct = (missing / len(airbnb) * 100).round(2)
print("결측값 개수:")
print(missing[missing > 0] if missing.sum() > 0 else pd.Series(dtype=int))
print()
print("결측값 비율(%):")
print(missing_pct[missing_pct > 0] if missing_pct.sum() > 0 else "→ 결측값 없음")
print()
print("[AI 연결]")
print("→ 결측값이 있으면 AI 모델이 오류를 일으키거나 성능이 급락한다.")
print("→ 따라서 EDA에서 결측값을 반드시 먼저 파악해야 한다.")
````

### Ping 8 — EDA: 조건 필터링

원하는 조건의 데이터만 골라내는 방법을 배운다.

````python
# ─── Ping 8: 필터링 ───────────────────────────────────────────
print("=" * 55)
print("  [조건 1] 단일 조건: Manhattan 숙소만")
print("=" * 55)
manhattan = airbnb[airbnb['neighbourhood_group'] == 'Manhattan']
print(f"→ Manhattan 숙소 수: {len(manhattan):,} 개")
cols_show = ['name', 'neighbourhood_group', 'price']
available_cols = [c for c in cols_show if c in airbnb.columns]
print(manhattan[available_cols].head(3))
print()

print("─" * 55)
print("  [조건 2] 복합 조건: 100달러 미만 & Private room")
print("─" * 55)
cheap_private = airbnb[
    (airbnb['price'] < 100) &
    (airbnb['room_type'] == 'Private room')
]
print(f"→ 저가 개인실 수: {len(cheap_private):,} 개")
print(cheap_private[['name', 'room_type', 'price']].head(3))
print()

print("─" * 55)
print("  [조건 3] 여러 값 중 하나: .isin()")
print("─" * 55)
bk_or_qn = airbnb[airbnb['neighbourhood_group'].isin(['Brooklyn', 'Queens'])]
print(f"→ Brooklyn 또는 Queens 숙소: {len(bk_or_qn):,} 개")
print()

print("[AI 연결]")
print("→ 필터링 = 훈련 데이터에서 이상치(outlier)를 제거하는 과정")
print("  예: price > 0 조건으로 무효 데이터 제거")
print("→ 트랜스포머(Transformer) 어텐션 마스킹도 동일 원리")
````

### Ping 9 — EDA: 정렬, 집계, 시각화

데이터의 분포와 패턴을 시각적으로 확인한다.

````python
# ─── Ping 9: 정렬 & value_counts & 시각화 ──────────────────────
print("=" * 55)
print("  [1] value_counts(): 범주 빈도")
print("=" * 55)
print("▷ 동네별 숙소 수:")
print(airbnb['neighbourhood_group'].value_counts())
print()
print("▷ 방 유형별 숙소 수:")
print(airbnb['room_type'].value_counts())
print()

print("─" * 55)
print("  [2] 가격 기준 정렬")
print("─" * 55)
print("▷ 가장 비싼 숙소 TOP 5:")
top5 = airbnb.sort_values('price', ascending=False).head(5)
print(top5[['name', 'neighbourhood_group', 'price']])
print()

print("─" * 55)
print("  [3] groupby(): 그룹별 통계")
print("─" * 55)
print("▷ 동네별 평균 가격:")
grp = airbnb.groupby('neighbourhood_group')['price'].mean().sort_values(ascending=False)
print(grp.round(2))
print()

print("─" * 55)
print("  [4] 시각화")
print("─" * 55)
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
fig.suptitle('에어비앤비 NYC EDA', fontsize=14, fontweight='bold')

# 동네별 숙소 수
ng_count = airbnb['neighbourhood_group'].value_counts()
ng_count.plot(kind='bar', ax=axes[0], color='#5B8DB8', edgecolor='white', linewidth=0.7)
axes[0].set_title('동네별 숙소 수', fontsize=12, fontweight='bold')
axes[0].set_xlabel('동네')
axes[0].set_ylabel('숙소 수')
axes[0].tick_params(axis='x', rotation=30)
for p in axes[0].patches:
    axes[0].annotate(f'{int(p.get_height()):,}',
                     (p.get_x() + p.get_width()/2., p.get_height()),
                     ha='center', va='bottom', fontsize=9)

# 방 유형별 비율
rt_count = airbnb['room_type'].value_counts()
rt_count.plot(kind='pie', ax=axes[1], autopct='%1.1f%%',
              colors=['#5B8DB8', '#E07B54', '#6BAA75'], startangle=90)
axes[1].set_title('방 유형별 비율', fontsize=12, fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
print()
print("[AI 연결]")
print("→ value_counts() = 분류 문제에서 클래스 불균형 확인 도구")
print("→ 불균형 클래스는 AI 모델 성능 저하의 주요 원인이다.")
````

## 전체 개념 정리

| 개념 | 핵심 메서드 | AI 연결 포인트 |
|:---|:---|:---|
| **시리즈**(Series) | `pd.Series()`, `.iloc`, `.loc` | 특성(feature) 벡터 |
| **데이터프레임**(DataFrame) | `pd.DataFrame()`, `.shape`, `.dtypes` | 훈련 데이터셋 |
| **데이터 불러오기** | `pd.read_csv()` | 원본 데이터 입력 |
| **기본 탐색** | `.head()`, `.info()`, `.describe()` | 데이터 품질 파악 |
| **결측값 확인** | `.isnull().sum()` | 모델 오류 예방 |
| **필터링** | 불리언 인덱싱, `.isin()` | 이상치 제거, 샘플 선택 |
| **집계** | `.groupby()`, `.value_counts()` | 클래스 불균형 확인 |
| **정렬** | `.sort_values()` | 중요 특성 탐색 |
| **정규화** | (x - min) / (max - min) | 딥러닝 전처리 기본 |

## AI 파이프라인에서 판다스의 위치

```
원본 데이터 (CSV / Excel / DB)
        ↓  pd.read_csv()
판다스: 탐색 (EDA)
        ↓  .isnull(), .describe(), .value_counts()
판다스: 전처리 (결측값, 이상치, 인코딩)
        ↓  .fillna(), 불리언 필터링
판다스 → 넘파이: df.values / df.to_numpy()
        ↓  torch.tensor()
파이토치 텐서 → AI 모델 훈련
```
