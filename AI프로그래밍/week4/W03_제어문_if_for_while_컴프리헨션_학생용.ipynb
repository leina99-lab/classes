# PART 1. 조건문 if

**조건문**(Conditional Statement)이란, 주어진 조건의 참/거짓에 따라 프로그램의 실행 흐름을 분기하는 구문이다.  
일상에서도 우리는 끊임없이 조건 판단을 한다.

→ "만약 비가 오면 우산을 가져간다. 그렇지 않으면 그냥 나간다."

파이썬에서는 이러한 판단을 `if`, `elif`, `else` 키워드로 표현한다.

## 1.1 기본 if문

가장 단순한 형태의 조건문이다. 조건이 **참**(True)일 때만 들여쓰기된 블록을 실행한다.

```python
if 조건식:
    실행할 문장
```

→ 조건식 뒤에 **콜론**( `:` )을 반드시 붙여야 한다.  
→ 실행할 문장은 **들여쓰기**(indent, 보통 공백 4칸)로 구분한다.  
→ 파이썬은 중괄호 `{}` 대신 **들여쓰기**로 코드 블록을 구분하므로, 들여쓰기가 매우 중요하다.

### 예제 1-1: 양수 판별

````python
x = 5
if x > 0:
    print(f'{x}는 양수이다.')
````

### 예제 1-2: 사용자 입력을 받아 100보다 큰지 판별

````python
# Colab/Jupyter에서 input()을 사용한다
# x = int(input('x의 값을 입력하세요: '))
# 여기서는 값을 직접 대입한다
x = 150
if x > 100:
    print('100보다 크다.')
````

### 예제 1-3: 문자열 비교

````python
food = '채식'
if food == '채식':
    print('당신은 채식을 좋아한다.')
````

### 예제 1-4: 들여쓰기 블록 — 여러 줄 실행

````python
age = 18
if age < 20:
    print('나이:', age)
    print('청소년 환영')
    print('청소년 할인')
````

위 코드에서 `age = 24`로 바꾸면 세 줄 모두 실행되지 않는다. 들여쓰기 블록 **전체**가 하나의 단위이기 때문이다.

````python
age = 24
if age < 20:
    print('나이:', age)
    print('청소년 환영')
    print('청소년 할인')

print('프로그램 종료')  # 이 줄은 if 블록 밖이므로 항상 실행된다
````

## 1.2 들여쓰기 주의사항

파이썬은 들여쓰기가 코드 블록을 구분하는 유일한 수단이다. 잘못된 들여쓰기는 `IndentationError`를 발생시킨다.

**올바른 예:**
```python
if x > 0:
    print('양수')   # 4칸 들여쓰기
```

**잘못된 예:**
```python
if x > 0:
print('양수')       # IndentationError!
```

````python
# 아래 코드를 실행하면 IndentationError가 발생한다
# x = 3
# if x == 3:
# print('3이다.')  # 들여쓰기가 없음!
````

## 1.3 if-else문

조건이 참이면 A를, 거짓이면 B를 실행하고 싶을 때 `else`를 사용한다.

```python
if 조건식:
    조건이 참일 때 실행할 문장
else:
    조건이 거짓일 때 실행할 문장
```

→ `else` 뒤에도 콜론( `:` )을 붙인다.  
→ `else`에는 별도의 조건식을 적지 않는다.

### 예제 1-5: 오전/오후 판별

````python
hour = 10
if hour < 12:
    print('오전이다.')
else:
    print('오후이다.')
````

### 예제 1-6: 음수/양수 판별 + 짝수/홀수 판별 (중첩 if)

## 1.4 if-elif-else문

조건이 3개 이상일 때 `elif`(else if의 줄임)를 사용한다.

```python
if 조건1:
    조건1이 참일 때 실행
elif 조건2:
    조건2가 참일 때 실행
elif 조건3:
    조건3이 참일 때 실행
else:
    모든 조건이 거짓일 때 실행
```

→ `elif`는 원하는 만큼 여러 개 사용할 수 있다.  
→ 위에서부터 순서대로 검사하며, **처음으로 참인 조건**의 블록만 실행하고 나머지는 건너뛴다.

### 예제 1-9: 학점 산출기

````python
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f'점수: {score}, 등급: {grade}')
````

### 예제 1-10: BMI(체질량지수) 판정

````python
h = 1.75   # 키(미터)
w = 70     # 몸무게(킬로그램)
BMI = w / (h ** 2)

print(f'BMI = {BMI:.1f}')

if BMI <= 18.5:
    print('저체중이다.')
elif BMI <= 22.9:
    print('정상 체중이다.')
elif BMI <= 24.9:
    print('비만 전 단계 체중이다.')
elif BMI <= 29.9:
    print('1단계 비만 체중이다.')
elif BMI <= 34.9:
    print('2단계 비만 체중이다.')
else:
    print('고도비만 체중이다.')
````

### 예제 1-11: 윤년 판별 (복합 조건식)

````python
year = 2024

is_leap_year = ((year % 4 == 0) and (year % 100 != 0)) or (year % 400 == 0)

if is_leap_year:
    print(f'{year}년은 윤년이다.')
else:
    print(f'{year}년은 평년이다.')
````

## 1.5 비교연산자와 논리연산자 정리

| 비교연산자 | 의미 |
|:---:|:---|
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |
| `==` | 같다 |
| `!=` | 같지 않다 |

| 논리연산자 | 의미 |
|:---:|:---|
| `and` | 둘 다 참이면 참 |
| `or` | 하나라도 참이면 참 |
| `not` | 참이면 거짓, 거짓이면 참 |
| `in` | 포함 여부 확인 |
| `not in` | 미포함 여부 확인 |

````python
# 논리연산자 예제
a = 3
b = 1

print('a == b and a >= b:', (a == b) and (a >= b))   # False and True → False
print('a >= b or a <= b:', (a >= b) or (a <= b))     # True or False → True
print('not a == b:', not (a == b))                    # not False → True
````

````python
# in / not in 연산자
fruits = ['사과', '바나나', '포도']
print('사과' in fruits)       # True
print('수박' in fruits)       # False
print('수박' not in fruits)   # True
````

# PART 2. 반복문 for

**반복문**(Loop)이란, 특정 작업을 여러 번 되풀이해서 수행하는 구문이다.  
컴퓨터는 인간과 달리 반복적인 작업을 실수 없이 빠르게 수행할 수 있다.

반복문에는 `for`문과 `while`문 두 종류가 있으며, `for`문은 **반복 횟수가 미리 정해져 있는 경우**에 주로 사용한다.

## 2.1 for문의 기본 구조

```python
for 변수 in 순회 가능한 객체:
    반복 실행할 문장
```

"순회 가능한 객체"에는 리스트, 문자열, `range()` 등이 올 수 있다.

### 예제 2-1: 기본 for문 — "Welcome" 5회 출력

````python
for i in range(5):
    print('Welcome to everyone!!')
````

### 예제 2-2: 루프 제어변수 i 활용

````python
for i in range(5):
    print(f'{i}번째 반복')
````

→ `range(5)`는 0, 1, 2, 3, 4를 차례로 생성한다. (0부터 시작하여 5 **미만**까지)  
→ 루프 제어변수 `i`를 사용하지 않을 경우, 언더스코어 `_`로 대체할 수 있다.

````python
for _ in range(3):
    print('반복 실행')
````

## 2.2 range() 함수 상세

`range()` 함수는 특정 구간의 정수 열(sequence)을 생성한다.

| 형태 | 의미 |
|:---|:---|
| `range(n)` | 0부터 n-1까지 (간격 1) |
| `range(start, stop)` | start부터 stop-1까지 (간격 1) |
| `range(start, stop, step)` | start부터 stop-1까지 (간격 step) |

### 예제 2-3: range() 다양한 사용법

````python
print('range(5)       →', list(range(5)))          # [0,1,2,3,4]
print('range(0, 5)    →', list(range(0, 5)))       # [0,1,2,3,4]
print('range(0, 5, 1) →', list(range(0, 5, 1)))    # [0,1,2,3,4]
print('range(0, 5, 2) →', list(range(0, 5, 2)))    # [0,2,4]
print('range(2, 5)    →', list(range(2, 5)))        # [2,3,4]
print('range(0, 10, 2)→', list(range(0, 10, 2)))   # [0,2,4,6,8]
print('range(1, 10, 2)→', list(range(1, 10, 2)))   # [1,3,5,7,9]
print('range(-2,-10,-2)→', list(range(-2, -10, -2))) # [-2,-4,-6,-8]
````

### 예제 2-4: range()를 for문에 활용

````python
# 2에서 4까지 출력
for i in range(2, 5):
    print(i, end=' ')
print()  # 줄바꿈

# 0, 2, 4, 6, 8 출력
for i in range(0, 10, 2):
    print(i, end=' ')
print()

# -2, -4, -6, -8 출력
for i in range(-2, -10, -2):
    print(i, end=' ')
````

## 2.3 반복문의 활용: 누적 합 계산

반복문에서 가장 기본적인 활용은 **누적 합**(cumulative sum)이다.  
변수를 0으로 초기화한 뒤, 반복할 때마다 값을 더해 나간다.

### 예제 2-5: 1부터 10까지의 합

````python
s = 0
for i in range(1, 11):
    s = s + i
    print(f'i={i}, 누적합={s}')

print(f'\n1부터 10까지의 합: {s}')
````

### 예제 2-6: 팩토리얼(factorial) 구하기

````python
# n! = 1 × 2 × 3 × ... × n
n = 10
s = 1
for i in range(2, n + 1):
    s = s * i
    print(f'i = {i}, s = {s}')

print(f'\n{n}! = {s}')
````

`while`문은 **조건이 참인 동안** 계속 실행하는 반복문이다.

# PART 4. break와 continue

반복문의 실행 흐름을 제어하는 두 가지 키워드이다.

| 키워드 | 기능 |
|:---:|:---|
| `break` | 반복문을 **즉시 종료**하고 루프를 빠져나간다 |
| `continue` | 현재 반복의 **나머지 부분을 건너뛰고** 다음 반복으로 넘어간다 |

## 4.1 break

### 예제 4-1: 모음을 만나면 중단

````python
st = 'Programming'

for ch in st:
    if ch in ['a', 'e', 'i', 'o', 'u']:
        break      # 모음을 만나면 즉시 종료
    print(ch)

print('The end')
````

→ `P`, `r`까지 출력된 후 `o`(모음)를 만나서 루프가 종료된다.

### 예제 4-2: while + break — 양수 입력 받기

````python
# 시뮬레이션: 입력값 목록으로 대체
inputs = [-10, 0, -8, 8]  # 순서대로 입력된다고 가정
idx = 0

n = -1
while n <= 0:
    n = inputs[idx]
    idx += 1
    print(f'입력값: {n}')

s = 0
for i in range(1, n + 1):
    s += i
print(f'1부터 {n}까지의 합은 {s}')
````

## 4.2 continue

### 예제 4-3: 모음을 건너뛰고 자음만 출력

````python
st = 'Programming'

for ch in st:
    if ch in ['a', 'e', 'i', 'o', 'u']:
        continue   # 모음이면 아래 print를 건너뛴다
    print(ch)

print('The end')
````

→ `break`와 달리, 루프 자체는 끝까지 계속 실행된다. 모음일 때만 `print(ch)`를 건너뛴다.

### 예제 4-4: 1~10에서 홀수만 출력 (continue)

````python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=' ')
````

### 예제 4-5: break vs continue vs pass 비교

````python
print('=== break ===')
for i in range(1, 11):
    if i % 2 == 0:
        break
    print(i, end=' ')

print('\n\n=== continue ===')
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=' ')

print('\n\n=== pass ===')
for i in range(1, 11):
    if i % 2 == 0:
        pass
    print(i, end=' ')
````

# PART 5. 리스트 컴프리헨션(List Comprehension)

**리스트 컴프리헨션**이란, for문과 조건문을 한 줄로 압축하여 리스트를 생성하는 파이썬만의 강력한 문법이다.  
기존 for문 + append() 방식보다 **간결하고 빠르며**, 파이썬 프로그래밍에서 매우 자주 사용된다.

→ AI/ML 분야에서 데이터 전처리를 할 때 리스트 컴프리헨션은 필수적으로 활용된다.

## 5.1 기본 문법

### for문 방식 vs 리스트 컴프리헨션 비교

**목표:** 0부터 4까지의 제곱수 리스트를 만든다.

**for문 방식:**
```python
squares = []
for i in range(5):
    squares.append(i ** 2)
```

**리스트 컴프리헨션:**
```python
squares = [i ** 2 for i in range(5)]
```

두 코드의 결과는 동일하지만, 리스트 컴프리헨션이 한 줄로 간결하게 표현된다.

**기본 구조:**
```python
[표현식 for 변수 in 순회 가능한 객체]
```

### 예제 5-1: 기본 리스트 컴프리헨션 — for문과 비교

````python
# === for문 방식 ===
squares_for = []
for i in range(5):
    squares_for.append(i ** 2)
print('for문 결과:', squares_for)

# === 리스트 컴프리헨션 ===
squares_comp = [i ** 2 for i in range(5)]
print('컴프리헨션 결과:', squares_comp)

# 결과 동일 확인
print('결과가 같은가?', squares_for == squares_comp)
````

### 예제 5-2: 다양한 기본 컴프리헨션

````python
# 1~10의 리스트
nums = [i for i in range(1, 11)]
print('1~10:', nums)

# 1~10의 2배 리스트
doubles = [i * 2 for i in range(1, 11)]
print('2배:', doubles)

# 문자열 리스트 → 대문자 변환
words = ['hello', 'world', 'python']
upper_words = [w.upper() for w in words]
print('대문자:', upper_words)
````

## 5.3 if-else를 포함한 리스트 컴프리헨션

조건에 따라 **다른 값**을 넣고 싶을 때는 `if-else`를 **표현식 위치**(for 앞)에 적는다.

**구조:**
```python
[참일때값 if 조건 else 거짓일때값 for 변수 in 순회객체]
```

→ 주의: 필터링용 `if`는 for 뒤에, 값 변환용 `if-else`는 for 앞에 온다!

### 예제 5-6: 짝수/홀수 라벨 붙이기

### 예제 5-7: 양수/음수/영 분류

````python
numbers = [3, -1, 0, 7, -5, 2, 0, -3]

# for문 방식
categories_for = []
for n in numbers:
    if n > 0:
        categories_for.append('양수')
    elif n < 0:
        categories_for.append('음수')
    else:
        categories_for.append('영')

# 컴프리헨션 (중첩 if-else)
categories_comp = ['양수' if n > 0 else ('음수' if n < 0 else '영') for n in numbers]

print('for문:', categories_for)
print('컴프리헨션:', categories_comp)
````

## 5.4 중첩 리스트 컴프리헨션

이중 for문도 리스트 컴프리헨션으로 표현할 수 있다.

**for문 방식:**
```python
result = []
for i in range(3):
    for j in range(3):
        result.append((i, j))
```

**리스트 컴프리헨션:**
```python
result = [(i, j) for i in range(3) for j in range(3)]
```

### 종합문제 1: 자동 판매기 프로그램

자동 판매기를 시뮬레이션하는 프로그램을 작성하라.  
투입 금액과 물건값(100원 단위)을 입력받아, 거스름돈을 500원 동전과 100원 동전의 개수로 계산하여 출력하라.

````python
money = 5000
price = 2600

change = money - price
coins_500 = change // 500
coins_100 = (change % 500) // 100

print(f'투입한 돈: {money}')
print(f'물건값: {price}')
print(f'거스름돈: {change}')
print(f'500원 동전의 개수: {coins_500}')
print(f'100원 동전의 개수: {coins_100}')
````

### 종합문제 2: 알파벳 다이아몬드 패턴

아래와 같은 패턴을 for문으로 출력하라:
```
    A
   BAB
  CBABC
 DCBABCD
EDCBABCDE
```

````python
n = 5  # 줄 수
for i in range(n):
    # 앞쪽 공백
    line = ' ' * (n - 1 - i)
    # 왼쪽 절반 (역순)
    for j in range(i, 0, -1):
        line += chr(65 + j)   # A=65
    # 중앙 A
    line += 'A'
    # 오른쪽 절반
    for j in range(1, i + 1):
        line += chr(65 + j)
    print(line)
````

### 종합문제 3: 원의 내부/외부 판별

중심이 원점 (0,0)이고 반지름이 5인 원이 있다. 점 (x, y)가 원의 내부에 있는지 외부에 있는지 판별하라.  
→ 힌트: 점과 원점 사이의 거리 = `(x**2 + y**2) ** 0.5`

````python
import math

test_points = [(2, 4), (7, 9), (3, 6)]

for x, y in test_points:
    distance = math.sqrt(x**2 + y**2)
    if distance <= 5:
        print(f'점 ({x}, {y}) → 원의 내부에 있다 (거리: {distance:.2f})')
    else:
        print(f'점 ({x}, {y}) → 원의 외부에 있다 (거리: {distance:.2f})')
````

def is_prime(n):
    if n < 2:

    for i in range(2, int(n**0.5) + 1):

    return True

primes = [n for n in range(2, 101) if is_prime(n)]

|:---|:---|
